```python
#function creation and function call


def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Division by zero is not allowed"
    return a / b

# Take two numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Display the results
print("Addition:", add(num1, num2))
print("Subtraction:", subtract(num1, num2))
print("Multiplication:", multiply(num1, num2))
print("Division:", divide(num1, num2))
```

```python
#temperature conversion
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9
def celsius_to_kelvin(celsius):
    return celsius + 273.15
string = input("Enter the temperature conversion you want to perform (e.g., 'C to F', 'F to C', 'C to K'): ")
string = string.upper().strip()
if string == "C TO F":
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = celsius_to_fahrenheit(celsius)
    print(f"{celsius}°C is equal to {fahrenheit}°F")
elif string == "F TO C":
    fahrenheit = float(input("Enter temperature in Fahrenheit: "))
    celsius = fahrenheit_to_celsius(fahrenheit)
    print(f"{fahrenheit}°F is equal to {celsius}°C")
elif string == "C TO K":
    celsius = float(input("Enter temperature in Celsius: "))
    kelvin = celsius_to_kelvin(celsius)
    print(f"{celsius}C is equal to {kelvin}K")
else:
    print("INVALID CHOICE")
```

```python
#prime check
def check_prime(num):
    if num > 1:
        for i in range(2, int(num/2)+1):
            if (num % i) == 0:
                return False
        else:
            return True
    else:
        return False
float_num = float(input("Enter a number to check if it is prime or not: "))
if float_num.is_integer():
    num = int(float_num)
    if check_prime(num):
        print(num, "is a prime number")
    else:
        print(num, "is not a prime number")
```

```python
#string reverse
def str_rev(str):
    return str[::-1]
str = input("Enter a string to reverse: ")
reversed_str = str_rev(str)
print("The reversed string is:", reversed_str)
```
