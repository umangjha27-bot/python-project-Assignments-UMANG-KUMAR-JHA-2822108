```python
#Calculating Factorial of 7 with a Loop.
fact = 1
for i in range(1, 8):
    fact *= i
print("Factorial of 7 is:", fact)
```

```python
#Check whether a user given input year is leap year or not.
yr = int(input("Enter a year: "))
if (yr % 4 == 0):
    print(yr, "is a leap year")
else:
    print(yr, "is not a leap year")
```

```python
#Write a program that takes a string such as "remember" and reverses it.
str = input(str("Enter a string: "))
rev = str[::-1]
print("Original string is:", str)
print("Reversed string is:", rev)
```
