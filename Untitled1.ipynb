```python
#sum till user enters 0
total = 0
while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    else:
        total += num
print("The total sum is:", total)
```

```python
#number reverse
i = int(input("Enter a number: "))
rev = 0
if i<0:
    print("Please enter a positive number")
else:
    while i>0:
        dig = i%10
        rev = rev*10 + dig
        i //= 10
    print("The reverse of the number is:", rev)
```

```python
#sum of digits
i = int(input("ENter a number:"))
s = 0
if i < 0:
    print("Please enter a positive number")
else:
    while i > 0:
        dig = i % 10
        s += dig
        i //= 10
    print("The sum of digits is:", s)
```
